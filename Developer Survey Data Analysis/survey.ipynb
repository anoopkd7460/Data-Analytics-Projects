#### Finding outliers in the dataset

```python
# Maximum Salary the developer have
max_value = data['ConvertedCompYearly'].max()
# this data is fake bcz no any developer can get this salary
max_value
```

```python
# second highest salary or less than the highest salary
second_highest=data[data['ConvertedCompYearly']<max_value]['ConvertedCompYearly']
second_highest
```

```python
# sort the salary in descending order
second_highest.sort_values(ascending=False).head(20)
```

```python
# get how many people making salary more than 95% of the people and we check this for different-2 quantile values

data['ConvertedCompYearly'].quantile(0.95) # first check
data['ConvertedCompYearly'].quantile(0.998) # second check
max_salary = data['ConvertedCompYearly'].quantile(0.999)  # third check
```

```python
# check all the rows which have outliers (these are outliers)
data[data['ConvertedCompYearly']>max_salary]
```

```python
# remove those salary data whose salary>max_salary which i have set previously and store in original data
data = data[data['ConvertedCompYearly']<max_salary]
```

```python
data['ConvertedCompYearly'].max()
```

```python
# details about that person whose salary is maximum in updated dataset
data[data['ConvertedCompYearly']==data['ConvertedCompYearly'].max()]
```

```python
# the person who get least amount of salary
data['ConvertedCompYearly'].min()
```

```python
# getting those values who are actually employed
emp_data = data[(data['Employment']=='Employed') & data['ConvertedCompYearly']!=np.nan]
emp_data['ConvertedCompYearly'].min()
```

```python
# By the help of above code we get that we have the outlier at the lower side as well
min_salary = emp_data['ConvertedCompYearly'].quantile(0.02)
min_salary
```

```python
# Plot this data
import matplotlib.pyplot as plt
percent_spread=data['EdLevel'].value_counts(normalize=True) * 100
percent_spread=percent_spread.sort_values()
percent_spread.plot(kind='barh')
plt.title('Distribution of Education Levels')
plt.xlabel('Percentage')
plt.ylabel("Education Level")
plt.show()
```

```python
# Calculate Offline Learning Percentage share
data['LearnCode'].value_counts() # This gives all the sources separated by colen
learn_methods = data['LearnCode'].str.split(';').explode() # describes whole data clearly
method_counts=learn_methods.value_counts() # all the responses separated out in a single object
method_percentages = (method_counts/89184)*100
method_percentages
```

```python
import matplotlib.pyplot as plt
method_percentages=method_percentages.sort_values()
method_percentages.plot(kind='barh')
plt.title('Learning methods')
plt.xlabel('Percentage')
plt.ylabel('Learn Code Methods')
for index,value in enumerate(method_percentages):
    plt.text(value,index,f'{value:.2f}%',va='center')
plt.show()
```

```python
# Calculate online Learning Percentage value
data['LearnCodeOnline'].value_counts()
# split value
data['LearnCodeOnline'].str.split(';') # this create a list of that
online_methods=data['LearnCodeOnline'].str.split(';').explode() # break the list and describe every learning method
online_method_counts=online_methods.value_counts()
online_method_percentages=(online_method_counts/89184)*100
online_method_percentages
```

```python
import matplotlib.pyplot as plt
online_method_percentages = online_method_percentages.sort_values()
online_method_percentages.plot(kind='barh')
plt.title('Learning methods')
plt.xlabel('Percentage')
plt.ylabel('Learn code Methods')
for index,value in enumerate(online_method_percentages):
    plt.text(value,index,f'{value:.2f}%',va='center')
plt.show()
```

```python
# Caculating Learning Platform's Percentage share
online_platform_responses=data[data['LearnCodeCoursesCert']!=np.nan]
response_count=online_platform_responses.count() # this code gives total count of different-2 platforms
response_count=online_platform_responses['LearnCodeCoursesCert'].count() # give only that column count

online_platform = data['LearnCodeCoursesCert'].str.split(';').explode()
online_platform_counts = online_platform.value_counts()
online_platform_percentages = (online_platform_counts/response_count) *100
online_platform_percentages
```

```python
import matplotlib.pyplot as plt
online_platform_percentages = online_platform_percentages.sort_values()
online_platform_percentages.plot(kind='barh')
plt.title('Learning methods')
plt.xlabel('Percentage')
plt.ylabel('Online Platform Percentages')
for index,value in enumerate(online_platform_percentages):
    plt.text(value,index,f'{value:.2f}%',va='center')
plt.show()
```
