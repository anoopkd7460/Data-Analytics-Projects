1. answering questions about the branch
2. which location has highest and the lowest sales?
3. Represent the sales on a bar chart, also show the market share for
4. each location using a pie chart

```python
location_list = sales.groupby('Location')
```

```python
# prints all the locations which are in the location_list
location = [x for x,y in location_list]
location
```

```python
# represent the branches vs sales
sales.groupby('Location').sum()['Total']
```

```python
import matplotlib.pyplot as plt
plt.bar(location,sales.groupby('Location').sum()['Total'])
plt.show()
```

```python
plt.pie(sales.groupby('Location').sum()['Total'],labels=location,autopct='%1.1f%%')
plt.show()
```

```python
# Which location has more female customers and which location has more male
location_sales = sales.groupby(['Gender','Location']).count()['Invoice ID']
location_sales # this prints data as raw format
```

```python
unstacked_sales = location_sales.unstack(level=0)
unstacked_sales
```

```python
# this bar represents male and female shopers
unstacked_sales.plot(kind='bar')
```

1. What days of the month make most sales?

```python
sales['Date']
```

```python
# this to_datetime give us object to datatime format
datetime = pd.to_datetime(sales['Date']) # this give us all the dates
datetime.dt.day # this gives all the days
datetime.dt.month # this gives all the months of the year
datetime.dt.year # this gives year
```

```python
# we have to save all the days as a new column in the sales data
sales['Day'] = datetime.dt.day
sales.head(5) # day column has been created
```

```python
# gives some of total amount spend on a particular day
day_sales = sales.groupby('Day').sum()['Total']
day_sales
```

```python
day_sales.plot()
plt.show()
```

1. Which months make the most sales?

```python
# this adds new column in sales dataset
sales['Month'] = datetime.dt.month
sales.head(5)
```

```python
sales_by_month = sales.groupby('Month').sum()['Total']
```

```python
sales_by_month.plot(kind='bar')
plt.grid()
plt.show()
```

What hours make the most sales?

```python
sales['Time']
```

```python
# converts object to hour and save column to sales dataset
sales['Hour'] = pd.to_datetime(sales['Time']).dt.hour
sales.head(5)
```

```python
sales_by_hour = sales.groupby('Hour').sum()['Total']
sales_by_hour
```

```python
sales_by_hour.plot(grid=True)
plt.show()
```

1. What times do people make more epayments vs what time people make more cash payments?
